## 2.1 Dasar Teori
Conditional atau percabangan merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk membuat keputusan berdasarkan kondisi tertentu. Dengan adanya conditional, program dapat menentukan perintah yang akan dijalankan sesuai dengan hasil dari suatu kondisi. Jika kondisi terpenuhi, program akan menjalankan perintah tertentu, sedangkan jika tidak terpenuhi, program dapat menjalankan perintah lainnya.

Python menyediakan beberapa pernyataan conditional, yaitu if, elif, dan else. Pernyataan if digunakan untuk memeriksa kondisi pertama. Jika kondisi tersebut bernilai benar (True), maka perintah di dalamnya akan dijalankan. Pernyataan elif digunakan untuk memeriksa kondisi lain apabila kondisi sebelumnya tidak terpenuhi, sedangkan else digunakan sebagai pilihan terakhir ketika seluruh kondisi sebelumnya bernilai salah (False).

Conditional biasanya menggunakan operator perbandingan dan operator logika untuk menentukan suatu kondisi. Operator perbandingan meliputi ==, !=, >, <, >=, dan <=, sedangkan operator logika meliputi and, or, dan not. Penggunaan operator tersebut memungkinkan program melakukan pemeriksaan terhadap nilai atau beberapa kondisi sekaligus.

### 2.2.5 Program multi_if2_2611531011
</div>

```python
# program ini menggunakan fungsi input()
# program menghitung diskon belanja

# input dari user
total_belanja_1011 = float(input("masukkan total belanja (Rp): "))

# input status member (mengecek apakah user mengetik y atau t)
input_member_1011 = input("apakah anda member (y/t): ").strip().lower()
is_member_1011 = input_member_1011 in ["y", "t"]

# input status kode promo (mengecek apakah user mengetik y atau t)
input_promo_1011 = input("apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_1011 = input_promo_1011 in ["y", "t"]

total_diskon_persen_1011 = 0

#multi if terpisah: setiaap kondisi diperiksa independen
#diskon bisa di tumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus 

if total_belanja_1011 > 1000000:
    total_diskon_persen_1011 += 10 #diskon belanja besar

if is_member_1011:
    total_diskon_persen_1011 += 5 #diskon member

if kode_promo_valid_1011:
    total_diskon_persen_1011 += 15 #diskon voucher

# menghitung nominal diskon dan total bayar
nominal_diskon_1011 = total_belanja_1011 * (total_diskon_persen_1011 / 100)
total_bayar_1011 = total_belanja_1011 - nominal_diskon_1011

# output hasil
print("\n--- rincian pembayaran ---")
print(f"total diskon : {total_diskon_persen_1011}% (Rp {nominal_diskon_1011:,.0f})")
print(f"total bayar : Rp {total_bayar_1011:,.0f}")

print(f"total diskon yang anda dapatkan adalah {total_diskon_persen_1011}%")
# output total diskon yang anda dapat: 30% jika belanja > 1 juta, member, dan kode promo valid
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

Program tersebut merupakan program untuk menghitung diskon belanja berdasarkan total belanja, status member, dan validitas kode promo. Pengguna memasukkan total belanja serta menjawab beberapa pertanyaan dengan pilihan y atau t.

Program menggunakan multi if, sehingga setiap kondisi diperiksa secara terpisah. Jika beberapa kondisi terpenuhi, diskon akan diakumulasikan. Diskon yang diberikan yaitu 10% untuk belanja di atas Rp1.000.000, 5% untuk member, dan 15% untuk kode promo yang valid. Dengan demikian, pengguna dapat memperoleh diskon maksimal 30% jika semua kondisi terpenuhi.

Setelah total diskon diperoleh, program menghitung nominal diskon dan total pembayaran yang harus dibayar. Hasil akhirnya menampilkan persentase diskon, jumlah potongan harga, dan total pembayaran.

Program tersebut digunakan untuk mengonversi nilai angka menjadi nilai huruf menggunakan struktur percabangan if, elif, dan else. Pengguna memasukkan nilai angka, kemudian program memeriksa nilai tersebut berdasarkan rentang yang telah ditentukan.

Jika nilai 81 atau lebih, program menghasilkan A. Nilai 70–80 menghasilkan B, nilai 60–69 menghasilkan C, nilai 50–59 menghasilkan D, sedangkan nilai di bawah 50 menghasilkan E.

Program menggunakan elif agar kondisi diperiksa secara berurutan dan hanya satu nilai huruf yang ditampilkan sesuai dengan nilai yang dimasukkan. Dengan demikian, program dapat melakukan pengelompokan nilai secara otomatis berdasarkan kondisi yang telah ditentukan.


### 2.2.7 Program match_case1_2611531011
</div>

```python
#program ini menggunakan fungsi input ()
#program konversi nilai angka ke nama bulan

bulan_1011 = int(input("Masukkan nilai angka bulan (1-12): "))

match bulan_1011:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("angka tidak valid")
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

Program tersebut digunakan untuk mengonversi angka menjadi nama bulan menggunakan struktur match-case pada Python. Pengguna memasukkan angka dari 1 sampai 12, kemudian program mencocokkan angka tersebut dengan case yang sesuai.

Setiap case mewakili satu bulan, misalnya angka 1 menghasilkan Januari, angka 2 menghasilkan Februari, dan seterusnya hingga angka 12 menghasilkan Desember. Jika pengguna memasukkan angka di luar rentang 1–12, maka case _ akan dijalankan dan menampilkan pesan “angka tidak valid”.

Struktur match-case digunakan sebagai alternatif percabangan ketika terdapat banyak pilihan nilai yang perlu dicocokkan.

Program diawali dengan meminta data pengunjung berupa nama, umur, kepemilikan SIM C, dan jumlah tiket. Data umur dan jumlah tiket diubah menjadi tipe int, sedangkan jawaban kepemilikan SIM C diproses menggunakan .strip(), .lower(), dan [0] agar lebih mudah digunakan dalam percabangan.

2. Validasi Jumlah Tiket <br>
Program menggunakan if tunggal untuk memeriksa jumlah tiket. Jika jumlah tiket kurang dari atau sama dengan 0, program memberikan peringatan bahwa kuota tiket tidak valid.

3. Pemilihan Paket Wahana <br>
Pemilihan wahana menggunakan struktur match-case. Terdapat lima pilihan wahana dengan harga yang berbeda, mulai dari Safari Rimba hingga All-Access VIP. case _ digunakan sebagai kondisi default apabila pengguna memasukkan nomor paket yang tidak tersedia.

4. Validasi Kelayakan Wahana <br>
Program menggunakan if-elif-else dengan operator logika and dan != untuk menentukan kelayakan pengunjung. Khusus paket Motor ATV Ekstrim, umur dan kepemilikan SIM C menjadi pertimbangan. Untuk paket lainnya, program memeriksa apakah pengunjung telah berusia minimal 10 tahun. Seluruh kondisi dibuat tanpa menggunakan nested-if.

5. Perhitungan Diskon <br>
Program menghitung subtotal berdasarkan harga satuan dikalikan jumlah tiket. Selanjutnya digunakan beberapa if terpisah untuk memberikan diskon secara akumulatif. Diskon dapat berasal dari jumlah belanja, status member, kode promo, dan jumlah tiket. Jika beberapa syarat terpenuhi, seluruh diskon tersebut akan dijumlahkan.

6. Perhitungan Total Pembayaran <br>
Setelah persentase diskon diperoleh, program menghitung nominal diskon dan mengurangkannya dari subtotal untuk mendapatkan total pembayaran. Hasil perhitungan kemudian ditampilkan menggunakan f-string dengan format angka desimal agar lebih mudah dibaca.

7. Evaluasi Bonus dan Output <br>
Terakhir, program menggunakan if-else untuk menentukan bonus berdasarkan total pembayaran. Jika total bayar lebih dari Rp300.000, pengunjung mendapatkan Souvenir Gratis. Jika tidak, program menampilkan ucapan terima kasih. Program kemudian menampilkan seluruh rincian transaksi dan pesan “Program Selesai”.

<h1 align="center">BAB III<br>KESIMPULAN</h1>

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa conditional merupakan salah satu konsep penting dalam pemrograman Python yang digunakan untuk membuat keputusan berdasarkan kondisi tertentu. Struktur percabangan yang dipelajari meliputi `if`, `if-else`, `if-elif-else`, `multi-if`, serta `match-case`. Setiap struktur memiliki fungsi dan penggunaan yang berbeda sesuai dengan kebutuhan program.

Melalui praktikum ini, mahasiswa dapat memahami penerapan operator logika seperti `and` dan `!=` dalam menentukan suatu kondisi, serta penggunaan multi-if untuk menghasilkan keputusan yang dapat terakumulasi. Dengan memahami berbagai bentuk conditional tersebut, mahasiswa diharapkan mampu membuat program yang lebih terstruktur dan mampu mengambil keputusan berdasarkan data yang diberikan oleh pengguna. Penerapan conditional juga menjadi dasar penting untuk mempelajari konsep pemrograman yang lebih kompleks.
