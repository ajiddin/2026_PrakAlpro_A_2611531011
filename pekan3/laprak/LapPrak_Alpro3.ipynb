## 2.1 Dasar Teori
Python memiliki berbagai jenis operator yang digunakan untuk melakukan proses tertentu terhadap data dan variabel. Operator merupakan simbol atau kata khusus yang digunakan untuk melakukan operasi, seperti perhitungan, perbandingan, pemberian nilai, maupun pengolahan kondisi. Penggunaan operator menjadi salah satu dasar penting dalam pemrograman karena hampir setiap program membutuhkan operasi terhadap data.

1. Operator Aritmatika
Operator aritmatika digunakan untuk melakukan operasi perhitungan matematika pada Python. Operator yang termasuk di dalamnya adalah penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat (//), modulus atau sisa pembagian (%), dan perpangkatan (**). Operator aritmatika dapat digunakan pada berbagai tipe data numerik, seperti integer dan float. Penggunaannya banyak ditemukan dalam perhitungan nilai, transaksi, luas bangun, dan berbagai proses perhitungan lainnya.
2. Operator Perbandingan
Operator perbandingan digunakan untuk membandingkan dua nilai atau lebih. Hasil dari operasi perbandingan berupa nilai Boolean, yaitu True atau False. Operator yang digunakan meliputi sama dengan (==), tidak sama dengan (!=), lebih besar (>), lebih kecil (<), lebih besar atau sama dengan (>=), serta lebih kecil atau sama dengan (<=). Operator ini biasanya digunakan dalam percabangan dan pemeriksaan kondisi pada suatu program.
3. Operator Logika
Operator logika digunakan untuk menggabungkan atau memanipulasi beberapa kondisi. Python memiliki tiga operator logika utama, yaitu and, or, dan not. Operator and menghasilkan nilai benar apabila semua kondisi yang diperiksa bernilai benar, sedangkan or menghasilkan nilai benar apabila terdapat setidaknya satu kondisi yang benar. Sementara itu, not digunakan untuk membalik nilai Boolean dari suatu kondisi. Operator logika banyak digunakan dalam proses pengambilan keputusan pada program.
4. Operator Penugasan
Operator penugasan digunakan untuk memberikan nilai kepada variabel atau mengubah nilai yang telah tersimpan di dalam variabel. Operator yang paling dasar adalah =, sedangkan operator gabungan lainnya antara lain +=, -=, *=, /=, //=, %= dan **=. Operator tersebut dapat membuat penulisan program menjadi lebih sederhana, terutama ketika suatu variabel perlu diperbarui berdasarkan nilai sebelumnya.
5. Operator Bitwise
Operator bitwise digunakan untuk melakukan operasi pada tingkat bit dari suatu bilangan integer. Python menyediakan beberapa operator bitwise, yaitu AND (&), OR (|), XOR (^), NOT (~), left shift (<<), dan right shift (>>). Operasi ini bekerja dengan memproses representasi biner dari suatu bilangan. Operator bitwise biasanya digunakan dalam pengolahan data tingkat rendah, manipulasi bit, sistem komputer, dan beberapa kebutuhan pemrograman tertentu.
6. Operator Keanggotaan
Operator keanggotaan digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah objek seperti string, list, tuple, atau set. Python memiliki operator in dan not in. Operator in menghasilkan True apabila suatu nilai ditemukan di dalam objek, sedangkan not in menghasilkan True apabila nilai tersebut tidak ditemukan. Operator ini dapat digunakan untuk melakukan pemeriksaan data dengan cara yang sederhana.
7. Operator Identitas
Operator identitas digunakan untuk mengetahui apakah dua variabel mengacu pada objek yang sama di dalam memori. Python menyediakan operator is dan is not. Operator is menghasilkan True apabila kedua variabel mengacu pada objek yang sama, sedangkan is not menghasilkan True apabila keduanya tidak mengacu pada objek yang sama. Operator identitas berbeda dengan operator == yang digunakan untuk membandingkan nilai.

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("======================================")
print("3. OPERATOR BITWISE")
print("======================================")

angka1_1021 = int(input("Masukkan angka bitwise-1: "))
angka2_1021 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =",angka1_1021,"| biner =",bin(angka1_1021))
print("angka1 =",angka2_1021,"| biner =",bin(angka2_1021))

# Bitwise AND
hasil_1021 = angka1_1021 & angka2_1021
print("\nBitwise AND (&)")
print(angka1_1021,"&",angka2_1021,hasil_1021)
print("Biner hasil =",bin(hasil_1021))
print("Biner hasil (8 bit) =",format(hasil_1021,"08b"))

# Bitwise OR
hasil_1021 = angka1_1021 | angka2_1021
print("\nBitwise OR (|)")
print(angka1_1021,"|",angka2_1021,hasil_1021)
print("Biner hasil =",bin(hasil_1021))
print("Biner hasil (8 bit) =",format(hasil_1021,"08b"))

# Bitwise XOR
hasil_1021 = angka1_1021 ^ angka2_1021
print("\nBitwise XOR (^)")
print(angka1_1021,"^",angka2_1021,hasil_1021)
print("Biner hasil =",bin(hasil_1021))
print("Biner hasil (8 bit) =",format(hasil_1021,"08b"))

# Bitwise NOT
hasil_1021 = ~angka1_1021
print("\nBitwise NOT (~)")
print(angka1_1021,"~",angka2_1021,hasil_1021)
print("Biner hasil =",bin(hasil_1021))
print("Biner hasil (8 bit) =",format(hasil_1021,"08b"))

# Bitwise geser kiri
jumlah_geser_1021 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_1021 = angka1_1021 << jumlah_geser_1021
print("\nBitwise geser kiri (<<)")
print(angka1_1021,"<<",jumlah_geser_1021,"=",hasil_1021)
print("Biner hasil =",bin(hasil_1021))
print("Biner hasil (8 bit) =",format(hasil_1021,"08b"))

# Bitwise geser kanan
hasil_1021 = angka1_1021 >> jumlah_geser_1021
print("\nBitwise geser kanan (>>)")
print(angka1_1021,">>",jumlah_geser_1021,"=",hasil_1021)
print("Biner hasil =",bin(hasil_1021))
print("Biner hasil (8 bit) =",format(hasil_1021,"08b"))
```

Menghitung rata-rata harga setiap barang setelah diskon dengan membagi total pembayaran dengan jumlah barang.
16. Menampilkan hak akses pelanggan<br>
print(f"Member Access : {status_valid_1021}")<br>
print(f"Promo Access : {promo_valid_1021}")<br>
Menampilkan hak akses berdasarkan status member dan validitas kode promo.<br>
Jika pelanggan adalah member, Member Access bernilai True. Jika kode promo tersedia, Promo Access bernilai True.
17. Membuat kode transaksi<br>
kode_transaksi_1021 = int(status_valid_1021) << 0 | int(syarat_total_1021) << 1 | int(syarat_jumlah_1021) << 2 | int<br>(promo_valid_1021) << 3
Bagian ini mengubah nilai Boolean menjadi angka dan menyusunnya menggunakan operasi bitwise.<br>
True dikonversi menjadi 1, sedangkan False menjadi 0.<br>
Setiap kondisi ditempatkan pada posisi bit berbeda:<br>
Status member → bit 0<br>
Total belanja → bit 1<br>
Jumlah barang → bit 2<br>
Promo → bit 3<br>
Operator << digunakan untuk menggeser bit, sedangkan | digunakan untuk menggabungkan semua bit tersebut.
18. Membuat kode referensi<br>
kode_referensi_1021 = int(status_valid_1021) << 0 | int(syarat_total_1021) << 1 | int(promo_valid_1021) << 3<br>
Membuat kode referensi dengan menggunakan status member, total belanja, dan promo. Kondisi jumlah barang tidak dimasukkan ke dalam kode ini.
19. Menampilkan kode biner setiap kondisi<br>
print(f"{format(int(status_valid_1021) << 0,"04b")} | {format(int(syarat_total_1021) << 1,"04b")} | {format(int<br>(syarat_jumlah_1021) << 2,"04b")} | {format(int(promo_valid_1021) << 3,"04b")}")
Menampilkan masing-masing kondisi dalam bentuk biner 4 bit.<br>
Fungsi:<br>
format(nilai, "04b")<br>
digunakan untuk mengubah nilai menjadi biner dengan panjang minimal 4 bit.
20. Menampilkan kode transaksi<br>
print(f"Kode Biner   : {format(kode_transaksi_1021,"04b")}")<br>
print(f"Kode Desimal : {kode_transaksi_1021}")<br>
Menampilkan kode transaksi dalam bentuk biner dan desimal.
21. Pemeriksaan status member<br>
print(f"{format(kode_transaksi_1021,"04b")} & {format(int(status_valid_1021) << 0,"04b")}")<br>
Menampilkan operasi bitwise AND antara kode transaksi dan bit status member.<br>
print(f"Hasil Biner   : {format((kode_transaksi_1021) & int(status_valid_1021) << 0,"04b")}")<br>
print(f"Hasil Desimal : {(kode_transaksi_1021) & int(status_valid_1021) << 0}")<br>
Menghasilkan nilai berdasarkan pemeriksaan bit status member.
22. Pemeriksaan promo<br>
print(f"{format(kode_transaksi_1021,"04b")} & {format(int(promo_valid_1021) << 3,"04b")}")<br>
Menampilkan operasi AND untuk memeriksa bit promo.<br>
print(f"Hasil Biner   : {format((kode_transaksi_1021) & int(promo_valid_1021) << 3,"04b")}")<br>
print(f"Hasil Desimal : {(kode_transaksi_1021) & int(promo_valid_1021) << 3}")<br>
Menampilkan hasil pemeriksaan promo dalam bentuk biner dan desimal.
23. Menampilkan kode transaksi dan referensi<br>
print(f"Kode Transaksi : {format(kode_transaksi_1021,"04b")}")<br>
print(f"Kode Referensi : {format(kode_referensi_1021,"04b")}")<br>
Menampilkan kedua kode dalam bentuk biner 4 bit untuk dibandingkan.
24. Membandingkan kode dengan XOR<br>
print(f"{format(kode_transaksi_1021,"04b")} ^ {format(kode_referensi_1021,"04b")}")<br>
Operator ^ digunakan untuk membandingkan bit dari kode transaksi dan kode referensi. Bit yang berbeda akan menghasilkan 1.<br>
print(f"Hasil Biner   : {format((kode_transaksi_1021) ^ 11,"04b")}")<br>
Baris ini menghitung XOR dengan angka 11, bukan dengan kode_referensi_1021. Jika tujuannya membandingkan kedua kode, seharusnya:<br>
print(f"Hasil Biner   : {format(kode_transaksi_1021 ^ kode_referensi_1021, "04b")}")<br>
Sedangkan baris berikut memang membandingkan kedua kode:<br>
print(f"Hasil Desimal : {(kode_transaksi_1021) ^ (kode_referensi_1021)}")
25. Pergeseran kode transaksi<br>
print(f"{format(kode_transaksi_1021,"04b")} << 1")<br>
Menampilkan operasi pergeseran bit kode transaksi sebanyak 1 posisi ke kiri.<br>
print(f"Hasil Biner   : {format((kode_transaksi_1021) << 1,"04b")}")<br>
print(f"Hasil Desimal : {(kode_transaksi_1021) << 1}")<br>
Menggeser seluruh bit kode transaksi satu posisi ke kiri dan menampilkan hasilnya dalam bentuk biner dan desimal.
26. Menampilkan tanda selesai<br>
print("=== SELESAI ===")<br>
Menampilkan pesan bahwa program telah selesai dijalankan.
