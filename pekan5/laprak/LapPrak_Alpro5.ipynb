## 2.1 Dasar Teori
Conditional atau percabangan merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk membuat keputusan berdasarkan kondisi tertentu. Dengan adanya conditional, program dapat menentukan perintah yang akan dijalankan sesuai dengan hasil dari suatu kondisi. Jika kondisi terpenuhi, program akan menjalankan perintah tertentu, sedangkan jika tidak terpenuhi, program dapat menjalankan perintah lainnya.

Python menyediakan beberapa pernyataan conditional, yaitu if, elif, dan else. Pernyataan if digunakan untuk memeriksa kondisi pertama. Jika kondisi tersebut bernilai benar (True), maka perintah di dalamnya akan dijalankan. Pernyataan elif digunakan untuk memeriksa kondisi lain apabila kondisi sebelumnya tidak terpenuhi, sedangkan else digunakan sebagai pilihan terakhir ketika seluruh kondisi sebelumnya bernilai salah (False).

Conditional biasanya menggunakan operator perbandingan dan operator logika untuk menentukan suatu kondisi. Operator perbandingan meliputi ==, !=, >, <, >=, dan <=, sedangkan operator logika meliputi and, or, dan not. Penggunaan operator tersebut memungkinkan program melakukan pemeriksaan terhadap nilai atau beberapa kondisi sekaligus.

## 2.2 Kode Program

### 2.2.1 Program perulangan_for1_2611531011
</div>

```python
ulang_1011 = int(input("masukkan jumlah perulangan: "))

for i in range(ulang_1011):
    print(f"perulangan ke-{i}")
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

**Penjelasan:**

Program tersebut digunakan untuk melakukan perulangan sebanyak jumlah yang ditentukan oleh pengguna. Pengguna memasukkan jumlah perulangan yang kemudian disimpan dalam variabel ulang_1011.

Perulangan for menggunakan range(ulang_1011) untuk menentukan berapa kali proses dijalankan. Variabel i berfungsi sebagai penghitung perulangan, yang dimulai dari 0. Setiap kali perulangan dilakukan, program menampilkan nomor perulangan melalui print().

### 2.2.2 Program perulangan_for2_2611531011
</div>

```python
ulang_1011 = int(input("masukkan jum;ah perulangan: "))
print("perulangan ke-0 sampai ke-", ulang_1011-1)
for i_1011 in range (ulang_1011):
    print(i_1011, end=" ")
print()
print("perulangan ke-1 sampai ke-", ulang_1011)
for i_1011 in range(1,ulang_1011+1):
    print(i_1011, end=" ")
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

Program tersebut digunakan untuk menunjukkan dua bentuk penggunaan range() dalam perulangan for berdasarkan jumlah perulangan yang dimasukkan pengguna.

Pada perulangan pertama, range(ulang_1011) menghasilkan angka yang dimulai dari 0 hingga ulang_1011 - 1. Sedangkan pada perulangan kedua, range(1, ulang_1011 + 1) menghasilkan angka yang dimulai dari 1 hingga nilai ulang_1011.

Penggunaan end=" " membuat hasil perulangan ditampilkan dalam satu baris dan dipisahkan dengan spasi. Fungsi print() setelah perulangan pertama digunakan untuk membuat baris baru sebelum menampilkan hasil perulangan berikutnya.

### 2.2.3 perulangan_for3_2611531011
</div>

```python
ulang_1011 = int(input("masukkan jumlah perulangan: "))

jumlah_1011=0
for i_1011 in range(1, ulang_1011+1):
    print(i_1011, end=" ")
    jumlah_1011 = jumlah_1011 + i_1011

    if i_1011<ulang_1011:
        print("+", end=" ")
    else:
        print("=", jumlah_1011, end=" ")
print()
print("jumlah =", jumlah_1011)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

Program tersebut digunakan untuk menjumlahkan seluruh bilangan dari 1 sampai jumlah perulangan yang dimasukkan pengguna. Variabel jumlah_1011 digunakan sebagai tempat untuk menyimpan hasil penjumlahan sementara.

Perulangan for menggunakan range(1, ulang_1011 + 1) sehingga angka yang diproses dimulai dari 1 hingga nilai yang dimasukkan. Setiap angka kemudian ditambahkan ke jumlah_1011.

Program juga menggunakan conditional if-else untuk menampilkan tanda + di antara angka. Ketika mencapai angka terakhir, program menampilkan tanda = beserta hasil penjumlahannya.

### 2.2.4 Program nested_for1_2611531011
</div>

```python
batas_1011 = int(input("masukkan nilai batas: "))
for i_1011 in range(1, batas_1011 + 1):
    for j_1011 in range(1, (-1 * i_1011 + batas_1011) + 1):
        print(".", end=" ")
    print(i_1011)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

Program tersebut digunakan untuk membuat pola angka dengan jumlah titik yang semakin berkurang pada setiap baris. Pengguna memasukkan nilai batas yang menentukan jumlah baris pola.

Program menggunakan perulangan for bersarang. Perulangan pertama menggunakan i_1011 untuk menentukan nomor baris dari 1 hingga nilai batas. Perulangan kedua menggunakan j_1011 untuk mencetak titik (.), dengan jumlah titik yang berkurang setiap barisnya.

Setelah titik selesai dicetak, print(i_1011) digunakan untuk menampilkan angka pada akhir baris. Dengan demikian, semakin besar nomor baris, jumlah titik akan semakin sedikit.


### 2.2.5 Program nested_for2_2611531011
</div>

```python
batas_1011 = int(input("masukkan nilai batas: "))
for i_1011 in range(1, batas_1011 + 1):
    for j_1011 in range(1, batas_1011 + 1):
        print("*", end="")
    print() #pindah ke baris berikutnya
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

Program tersebut digunakan untuk membuat pola persegi dari karakter * berdasarkan nilai batas yang dimasukkan pengguna.

Perulangan for pertama menggunakan i_1011 untuk menentukan jumlah baris, sedangkan perulangan for kedua menggunakan j_1011 untuk menentukan jumlah bintang pada setiap baris. Karena keduanya menggunakan batas yang sama, jumlah baris dan kolom akan sama.

print("*", end="") digunakan agar bintang tetap dicetak pada baris yang sama. Setelah satu baris selesai, print() digunakan untuk berpindah ke baris berikutnya.

### 2.2.6 Program nested_for3_2611531011
</div>

```python
batas_1011 = int(input("masukkan nilai batas: "))
for i_1011 in range(batas_1011+1):
    for j_1011 in range(batas_1011+1):
        print(i_1011+j_1011, end=" ")
    print() #pindah ke baris berikutnya
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

Program tersebut digunakan untuk membuat pola angka berdasarkan hasil penjumlahan indeks baris dan kolom. Pengguna memasukkan nilai batas yang menentukan ukuran pola.

Program menggunakan nested loop. Perulangan pertama (i_1011) menentukan baris, sedangkan perulangan kedua (j_1011) menentukan kolom. Pada setiap posisi, program mencetak hasil dari i_1011 + j_1011.

print(..., end=" ") digunakan agar angka ditampilkan dalam satu baris dengan spasi, sedangkan print() digunakan untuk berpindah ke baris berikutnya.


### 2.2.7 Program nested_for4_2611531011
</div>

```python
tinggi_1011 = int(input("masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1011 %2 != 0:
    print("tinggi harus bilangan genap")
else:
    a_1011 = tinggi_1011
    c_1011 = a_1011
    lebar_1011 = (2*tinggi_1011) - 2

    for i_1011 in range(1, tinggi_1011 +1):
        b_1011 = c_1011 + 1

        for j_1011 in range(1, lebar_1011 + 1):

            #baris atas dan bawah
            if i_1011 == 1 or i_1011 == tinggi_1011:
                if j_1011 ==1 or j_1011 == lebar_1011:
                    print("#", end="")
                else:
                    print("=", end="")
            #baris isi
            else:
                if j_1011 == 1 or j_1011 == lebar_1011:
                    print("|", end="")
                else:
                    if j_1011 == c_1011:
                        print("<", end="")
                    elif j_1011 == b_1011:
                        print(">", end="")
                    elif j_1011 == (lebar_1011 - c_1011):
                        print("<", end="")
                    elif j_1011 == (lebar_1011 - c_1011 + 1):
                        print(">", end="")
                    elif j_1011>b_1011 and j_1011<(lebar_1011 - c_1011):
                        print(".", end="")
                    else:
                        print(" ", end="")
        print()

        #logika asli java
        a_1011 -= 2

        if a_1011 <= 0:
            c_1011 = (-a_1011) + 2
        else:
            c_1011 = a_1011
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

#### 1. Input dan Validasi

* **Input Nilai:** Program menerima masukan variabel `tinggi_1011` dari pengguna dalam bentuk bilangan bulat (*integer*).
* **Validasi Ganjil/Genap:** Menggunakan operasi modulus `tinggi_1011 % 2 != 0` untuk memastikan input berupa bilangan genap. Jika input ganjil, program menampilkan pesan peringatan dan berhenti.

#### 2. Inisialisasi Variabel

* **`a_1011` dan `c_1011`:** Diinisialisasi dengan nilai `tinggi_1011`. Variabel ini menentukan posisi horizontal untuk mencetak karakter pola `<`, `>`, dan `.` pada baris tertentu.
* **`lebar_1011`:** Dihitung dengan rumus `(2 * tinggi_1011) - 2` untuk menentukan jumlah kolom dalam matriks cetak.

#### 3. Perulangan Luar (Baris)

Menggunakan perulangan `for i_1011 in range(1, tinggi_1011 + 1)` untuk mengontrol cetakan baris demi baris dari atas ke bawah.

* Di awal baris, nilai `b_1011` diperbarui menjadi `c_1011 + 1`.

#### 4. Perulangan Dalam (Kolom / Pencetakan Karakter)

Menggunakan perulangan `for j_1011 in range(1, lebar_1011 + 1)` untuk mencetak karakter pada setiap kolom:

* **Baris Atas dan Bawah (`i_1011 == 1` atau `i_1011 == tinggi_1011`):**
* Mencetak karakter `#` pada pojok kiri (`j_1011 == 1`) dan pojok kanan (`j_1011 == lebar_1011`).
* Mencetak karakter `=` di antara kedua pojok tersebut.


* **Baris Isi (Tengah):**
* **Batas Kiri & Kanan:** Mencetak karakter `|` pada kolom pertama (`j_1011 == 1`) dan kolom terakhir (`j_1011 == lebar_1011`).
* **Pola Simbol Dalam:**
* Mencetak `<` jika posisi kolom cocok dengan `c_1011` atau `(lebar_1011 - c_1011)`.
* Mencetak `>` jika posisi kolom cocok dengan `b_1011` atau `(lebar_1011 - c_1011 + 1)`.
* Mencetak titik `.` pada area di antara simbol `>` kiri dan `<` kanan (`j_1011 > b_1011` dan `j_1011 < (lebar_1011 - c_1011)`).
* Mencetak spasi `" "` untuk area kosong lainnya.





#### 5. Pembaruan Nilai Koordinat Pola

Di akhir setiap iterasi baris:

* Nilai `a_1011` dikurangi $2$ (`a_1011 -= 2`).
* Pengondisian `if a_1011 <= 0` digunakan untuk membalikkan pergerakan koordinat `c_1011` setelah melewati setengah tinggi pola, sehingga bentuk pola simetris antara bagian atas dan bagian bawah.

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

n_1011 = int(input("Masukkan ukuran skala jam pasir (N): "))

# Border atas
print("#", end="")

for i_1011 in range(4 * n_1011 + 5):
    print("=", end="")

print("#")

# Fase 1: Jam Pasir Atas
for baris_1011 in range(n_1011, 0, -1):
    print("| ", end="")

    # Spasi penyeimbang kiri
    for spasi_1011 in range(2 * (n_1011 - baris_1011)):
        print(" ", end="")

    # Deret angka menurun
    for angka_1011 in range(baris_1011, 0, -1):
        print(angka_1011, end=" ")

    # Poros kristal
    print("<*>", end="")

    # Deret angka menaik
    for angka_1011 in range(1, baris_1011 + 1):
        print(" ", end="")
        print(angka_1011, end="")

    # Spasi penyeimbang kanan
    for spasi_1011 in range(2 * (n_1011 - baris_1011)):
        print(" ", end="")

    print(" |")

# Fase 2: Poros titik pusat
print("| ", end="")

for spasi_1011 in range(2 * n_1011 + 1):
    print(" ", end="")

print("<*>", end="")

for spasi_1011 in range(2 * n_1011 + 1):
    print(" ", end="")

print(" |")

# Fase 3: Jam Pasir Bawah
for baris_1011 in range(1, n_1011 + 1):
    print("| ", end="")

    # Spasi penyeimbang kiri
    for spasi_1011 in range(2 * (n_1011 - baris_1011)):
        print(" ", end="")

    # Deret angka menurun
    for angka_1011 in range(baris_1011, 0, -1):
        print(angka_1011, end=" ")

    # Poros kristal
    print("<*>", end="")

    # Deret angka menaik
    for angka_1011 in range(1, baris_1011 + 1):
        print(" ", end="")
        print(angka_1011, end="")

    # Spasi penyeimbang kanan
    for spasi_1011 in range(2 * (n_1011 - baris_1011)):
        print(" ", end="")

    print(" |")

# Border bawah
print("#", end="")

for i_1011 in range(4 * n_1011 + 5):
    print("=", end="")

print("#")
```

Penjelasan Alur Logika Program (tugas5_1011.py)

1. Input dan Judul Program<br>
Program diawali dengan menampilkan judul “Program Jam Pasir Kristal Palindromik”. Selanjutnya, pengguna diminta memasukkan nilai n_1011 sebagai ukuran atau skala dari pola jam pasir. Nilai tersebut digunakan untuk menentukan jumlah baris dan ukuran pola yang akan dibuat.

2. Pembuatan Border Atas<br>
Bagian ini digunakan untuk membuat bingkai bagian atas pola. Karakter # digunakan sebagai ujung bingkai, sedangkan karakter = dicetak secara berulang menggunakan perulangan for. Jumlah karakter = disesuaikan dengan nilai n_1011, sehingga ukuran bingkai dapat mengikuti skala yang dimasukkan pengguna.

3. Fase 1 – Jam Pasir Bagian Atas<br>
Bagian ini membentuk pola jam pasir dari ukuran terbesar menuju ukuran terkecil menggunakan range(n_1011, 0, -1). Setiap baris memiliki spasi penyeimbang di bagian kiri dan kanan agar bentuk pola tetap simetris. Deretan angka dicetak secara menurun dari nilai baris_1011 menuju 1, kemudian terdapat simbol <*> sebagai poros atau pusat kristal. Setelah itu, angka dicetak kembali secara menaik dari 1 hingga baris_1011. Kombinasi tersebut menghasilkan pola angka yang bersifat palindromik atau memiliki susunan yang simetris.

4. Fase 2 – Poros Titik Pusat<br>
Bagian ini digunakan untuk membuat bagian tengah jam pasir. Program mencetak spasi dengan jumlah tertentu berdasarkan n_1011, kemudian menampilkan simbol <*> sebagai pusat pola. Spasi di kedua sisi digunakan untuk menjaga posisi simbol tetap berada di tengah bingkai.

5. Fase 3 – Jam Pasir Bagian Bawah<br>
Bagian ini membentuk pola bagian bawah dengan menggunakan perulangan dari 1 hingga n_1011. Struktur yang digunakan hampir sama dengan bagian atas, tetapi urutan barisnya dibalik sehingga pola kembali melebar dari bagian tengah menuju ukuran terbesar. Deretan angka menurun dan menaik serta simbol <*> membuat pola tetap simetris dan membentuk keseluruhan jam pasir.

6. Pembuatan Border Bawah<br>
Bagian terakhir digunakan untuk membuat bingkai bagian bawah. Sama seperti border atas, karakter # digunakan sebagai pembatas dan karakter = dicetak secara berulang menggunakan for. Dengan demikian, pola jam pasir berada di dalam bingkai yang lengkap.
